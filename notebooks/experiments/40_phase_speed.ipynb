# 수동 허브와 모델 중심의 구간 속도 비율

같은 시도에서 이륙→무릎 통과와 무릎 통과→접촉의 평균 수직 속도를 비교합니다. 이 비율은 앞 구간 속도를 뒤 구간 속도로 나눈 값입니다.

```python
from pathlib import Path
from datetime import datetime
import pandas as pd
import matplotlib.pyplot as plt
from wlcv.config import ROOT
from wlcv.annotation.plate_blur import REVIEW_FILE
from wlcv.evaluation.phase_speed import compare_speeds, model_plates

folder = ROOT / "outputs/phase_speed" / datetime.now().strftime("%Y%m%d-%H%M%S-%f")
folder.mkdir(parents=True)
plates = model_plates(sorted(pd.read_csv(REVIEW_FILE).video_id.unique()))
plates.to_csv(folder / "selected.csv", index=False)
```

### 세 시점과 표시 시각 확인

수동점이 없는 구간은 빈칸으로 둡니다. 모델의 위치로 채우면 모델 오차를 평가할 수 없습니다.

```python
points, trials = compare_speeds(plates)
points.to_csv(folder / "points.csv", index=False)
trials.to_csv(folder / "trials.csv", index=False)
display(points.round(2), trials.round(3))
```

```python
ax = trials.set_index("video_id")[["manual_ratio", "model_ratio"]].plot.bar(rot=25, ylabel="Early / late vertical speed")
ax.axhline(1, color="grey", linestyle="--")
ax.figure.tight_layout()
ax.figure.savefig(folder / "ratios.png", dpi=140)
plt.show()
```
