# 원판 없는 영상의 오탐 평가

학습에 쓰지 않은 사람의 영상을 끝까지 검수한 뒤 `data/metadata/negative_eval.csv`에 등록합니다. 열은 `video_id,person_id,filename`이며, 파일은 `WL_VIDEO_DIR`에서 찾습니다.

```python
from wlcv.config import RUNS
from wlcv.evaluation.negative import MANIFEST, evaluate_negatives
```

## 1. 평가할 실험과 목록 선택

모델에 딸린 `train_people.csv`에는 실제 학습 프레임을 제공한 사람이 기록되어 있습니다. 같은 사람의 다른 영상도 평가에서 제외합니다.

```python
weights = RUNS / "plate_baseline" / "weights" / "best.pt"
manifest = MANIFEST
conf = 0.25
```

## 2. 모든 프레임 검사

CPU에서 처음부터 끝까지 검사합니다. `fp_boxes`는 검출한 박스의 총수이고, `fp_frames`는 박스가 하나 이상 나온 프레임 수입니다.

```python
details = evaluate_negatives(weights, manifest, conf=conf)
if not details.empty:
    display(details)
    frames = details["frames"].sum()
    false_frames = details["fp_frames"].sum()
    print(f"전체 오탐 프레임 비율 {false_frames}/{frames} ({false_frames / frames:.1%})")
```
