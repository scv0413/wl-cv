# 스내치 동작 시점 기록하기

준비(ready), 전환(transition), 추진(drive), 받기(catch), 마무리(finish)에 해당하는 프레임 번호를 적습니다. 하프 동작(h) 영상은 앞의 세 이벤트를 기록합니다.

전환(`transition`)은 바가 무릎을 통과하는 순간을 표시합니다. 앞뒤 프레임을 보면서 그 지점에서 체간(몸통)이 본격적으로 세워지기 전인지 함께 확인합니다.


```python
import ipywidgets as widgets
from IPython.display import display
from wlcv.annotation.frame_view import video_picker
from wlcv.config import VIDEOS
from wlcv.annotation.events import phase_event_editor
```

## 1. 영상과 판독 차수 고르기

1차와 2차 기록은 따로 저장합니다. 2차 이벤트를 하나라도 저장하면 그 영상의 1차 기록은 읽기 전용이 됩니다. 영상이나 차수를 바꾼 후 판독기 셀을 다시 실행하세요.


```python
video = video_picker(VIDEOS)
pass_number = widgets.Dropdown(description="판독 차수", options=[1, 2])
display(video, pass_number)
```

## 2. 동작 시점 기록과 재열기

동작을 고르고 슬라이더로 해당 프레임을 찾은 뒤 **현재 프레임 기록**을 누릅니다. **기록 다시 읽기**로 저장값을 확인합니다. 첫 저장 때 원본 기록을 `data/annotations/personal_point_labels.csv`에 복사합니다.

```python
display(phase_event_editor(video.value, pass_number.value))
```
