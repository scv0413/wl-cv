# 스내치 동작 시점 기록하기

준비(ready), 전환(transition), 추진(drive), 받기(catch), 마무리(finish)에 해당하는 프레임 번호를 적습니다. 하프 동작(h) 영상은 앞의 세 이벤트를 기록합니다.


```python
import ipywidgets as widgets
from IPython.display import display
from wlcv.config import VIDEOS, ANNOTATIONS
from wlcv.annotation.events import phase_event_editor
from wlcv.annotation.storage import load_events
```

## 1. 영상과 판독 차수 고르기

1차와 2차 기록은 따로 저장합니다. 2차 이벤트를 하나라도 저장하면 그 영상의 1차 기록은 읽기 전용이 됩니다. 영상이나 차수를 바꾼 후 판독기 셀을 다시 실행하세요.


```python
videos = sorted(VIDEOS.glob("*.mp4"))
if not videos:
    raise FileNotFoundError(f"{VIDEOS}에 MP4 영상을 넣어 주세요.")
video = widgets.Dropdown(description="영상", options=[(p.name, p) for p in videos])
pass_number = widgets.Dropdown(description="판독 차수", options=[1, 2])
display(video, pass_number)
```

## 2. 판독기 열기

**동작**을 선택합니다. 기존 기록이 있으면 해당 프레임으로 이동합니다. 슬라이더와 방향키로 동작 시점을 찾은 뒤 **현재 프레임 기록**을 누르세요. 다른 이벤트도 같은 순서로 기록합니다. 현재 판단하기 어려운 이벤트는 비워 둘 수 있습니다.


```python
display(phase_event_editor(video.value, pass_number.value))
```

## 3. 기록 다시 확인하기

**기록 다시 읽기**를 누르거나 아래 셀을 실행하세요. 첫 저장 때 원본 `point_labels.csv`의 행을 개인 파일 `personal_point_labels.csv`로 복사합니다. 이후 같은 영상·이벤트·차수를 저장하면 그 기록의 프레임 번호만 바뀝니다.


```python
records = load_events(ANNOTATIONS, video.value.stem, pass_number.value)
for name, index in records.items():
    print(f"{name:12s} 프레임 {index}")
if not records:
    print("이 차수에 기록한 이벤트가 없습니다.")
```
