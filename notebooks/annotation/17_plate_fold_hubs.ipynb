# 3-fold 판독 · 가까운 원판의 허브 중심

6명의 양성 라벨 39장에서 **카메라에 가까운 쪽 원판의 중앙, 바벨 축이 원판을 통과하는 한 점**을 찍습니다. 원본 프레임과 내가 저장한 점만 보입니다. 이 좌표는 나중에 같은 프레임에서 65·95에폭의 허브 세로 오차를 비교할 때 사용합니다.

## 1. 판독할 선수와 장수 확인

Fold 1은 a06·c02의 11장, Fold 2는 b04·c05의 14장, Fold 3은 a01·b01의 14장입니다. c05의 7장도 전체 프레임의 허브 비교에 사용합니다. 원판 없는 b05·b06과 기존 검증 6명은 여기서 판독하지 않습니다.

```python
from IPython.display import display
from wlcv.annotation.fold_hubs import POINT_DIR, fold_targets, fold_hub_editor, progress_table

targets = fold_targets()
display(progress_table(targets))
print(f"전체 {len(targets)}장 · 선수 {targets.person_id.nunique()}명")
```

## 2. 원본에서 허브 직접 찍기

허브가 보이면 이미지를 **한 번 클릭**하세요. 십자와 원본 픽셀 좌표가 나타나면 저장된 상태입니다. 작은 원판은 **원본 크기로 보기**를 켜고, 잘못 찍으면 같은 프레임에서 다시 클릭하세요. 가림이나 잔상 때문에 중심 한 점을 정하기 어려우면 **판독 불가로 저장**을 누르세요. 이때 좌표는 비워 둡니다.

**다음 미판독**으로 이동하세요. 드롭다운에서 이미 저장한 장면을 다시 열 수 있습니다. 중단 후 첫 셀부터 실행하면 첫 미판독 프레임으로 돌아옵니다.

```python
display(fold_hub_editor(targets))
```

## 3. 판독 진행 확인

아래 셀은 CSV를 다시 읽습니다. 직접 확인과 판독 불가의 합이 39이고 미판독이 0이면 이번 판독이 끝납니다. 선수별 CSV에 저장한 원본 픽셀 좌표에서 y는 화면 아래쪽으로 커집니다.

```python
counts = progress_table(targets)
display(counts)
saved = int(counts["직접 확인"].sum() + counts["판독 불가"].sum())
print(f"{saved}/{len(targets)}장 저장 · 직접 확인 {counts['직접 확인'].sum()}장 · 판독 불가 {counts['판독 불가'].sum()}장")
print("저장 폴더", POINT_DIR)
```
