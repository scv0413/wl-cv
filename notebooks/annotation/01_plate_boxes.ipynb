# 원판에 박스 그리기

카메라에 가까운 쪽 원판의 외곽을 표시합니다. 프레임 번호는 0부터 셉니다.


```python
import ipywidgets as widgets
from IPython.display import display
from wlcv.config import VIDEOS, ANNOTATIONS
from wlcv.annotation.boxes import plate_box_editor
```

## 1. 영상 고르기

아래 목록에서 영상을 고른 뒤 **편집기 열기** 셀을 실행하세요. 영상을 바꿀 때는 현재 작업을 저장하고 같은 셀을 다시 실행합니다.


```python
videos = sorted(VIDEOS.glob("*.mp4"))
if not videos:
    raise FileNotFoundError(f"{VIDEOS}에 MP4 영상을 넣어 주세요.")
video = widgets.Dropdown(description="영상", options=[(p.name, p) for p in videos])
display(video)
```

## 2. 편집기 열기

1. **라벨 위치**로 이미 검수한 프레임을 고르거나 슬라이더로 다른 프레임을 찾습니다. 슬라이더에 초점을 두고 방향키를 누르면 한 프레임씩 이동합니다.
2. 카메라에 가까운 원판 하나를 골라 외곽의 한 모서리에서 반대 모서리까지 드래그합니다. 잘못 그리면 **마지막 박스 삭제**를 누릅니다.
3. **박스 저장**을 누릅니다. 원판이 없음을 직접 확인했으면 박스를 모두 지우고 **원판 없음 저장**을 누릅니다.
4. **다시 읽기**를 눌러 저장한 박스 수와 테두리를 확인합니다. 저장 전 변경은 이 버튼으로 취소할 수 있습니다.


```python
display(plate_box_editor(video.value))
```

## 3. 저장 파일 확인하기

`data/annotations/personal_plate_boxes/`에 개인 라벨이 생깁니다. 파일이 비어 있으면 원판이 없다고 검수한 프레임입니다. 편집기는 개인 파일을 먼저 찾고, 없으면 `plate_boxes/`의 원본 라벨을 읽습니다. 아래 셀로 최근에 저장한 개인 파일의 좌표를 확인하세요. 한 줄은 클래스 `0`, 중심 x, 중심 y, 너비, 높이이며 좌표와 크기는 프레임에 대한 비율입니다.


```python
folder = ANNOTATIONS / "personal_plate_boxes"
saved = sorted(folder.glob(f"{video.value.stem}_frame_*.txt"), key=lambda p: p.stat().st_mtime)
if saved:
    print(saved[-1].name)
    print(saved[-1].read_text().strip() or "원판 없음으로 검수했습니다.")
else:
    print("이 영상에서 저장한 개인 박스가 없습니다.")
```
