# 원판에 박스 그리기

카메라에 가까운 쪽 원판의 외곽을 표시합니다. 프레임 번호는 0부터 셉니다.


```python
from IPython.display import display
from wlcv.annotation.frame_view import video_picker
from wlcv.config import VIDEOS, ANNOTATIONS
from wlcv.annotation.boxes import plate_box_editor
```

## 1. 영상 선택

영상을 바꿀 때는 현재 라벨을 저장한 뒤 아래 편집기 셀을 다시 실행합니다.

```python
video = video_picker(VIDEOS)
display(video)
```

## 2. 원판 박스 저장과 재열기

1. **라벨 위치**에서 기존 프레임을 고르거나 슬라이더로 이동합니다.
2. 가까운 원판의 한 모서리에서 반대 모서리까지 드래그하고 **박스 저장**을 누릅니다. 잘못 그렸으면 **마지막 박스 삭제**로 지웁니다.
3. 원판이 없으면 빈 상태에서 **원판 없음 저장**을 누릅니다. **다시 읽기**로 현재 프레임에 저장된 테두리를 확인합니다.

```python
display(plate_box_editor(video.value))
```

## 3. 저장한 박스 좌표 확인

개인 파일이 있으면 편집기가 먼저 읽습니다. 아래는 이 영상에서 가장 최근에 저장한 파일입니다. 각 행은 `클래스 중심x 중심y 너비 높이`이며 좌표와 크기는 0~1 범위입니다.

```python
folder = ANNOTATIONS / "personal_plate_boxes"
saved = sorted(folder.glob(f"{video.value.stem}_frame_*.txt"), key=lambda p: p.stat().st_mtime)
if saved:
    print(saved[-1].name)
    print(saved[-1].read_text().strip() or "원판 없음으로 검수했습니다.")
else:
    print("이 영상에서 저장한 개인 박스가 없습니다.")
```
