# b02 원판 허브 중심 블라인드 판독

34·35·36번에서 카메라에 가까운 원판에 바벨 축이 관통하는 한 점을 찍는다. 아래 화면에는 원본과 내가 저장한 점만 표시된다. 번짐이나 가림으로 한 점을 정할 수 없으면 **판독 불가로 저장**한다. 세 장은 기존 검증 분할에 그대로 둔다.

```python
import ipywidgets as widgets
from IPython.display import display
from wlcv.annotation.hub_points import FRAMES, hub_editor, read_hubs

tabs = widgets.Tab(children=[hub_editor(index) for index in FRAMES])
for number, index in enumerate(FRAMES):
    tabs.set_title(number, f"{index}번")
display(tabs)
```

각 탭에서 원본을 클릭하면 즉시 저장된다. 같은 탭을 다시 클릭해 수정할 수 있다. 세 탭을 마치고 아래 셀을 실행한다.

```python
points = read_hubs()
display(points.sort_values("frame_index"))
print(f"{len(points)}/3장 저장 · 직접 확인 {points.visibility.eq('visible').sum()}장")
if len(points) == 3:
    print("32번 실험 노트북에서 학습 일정을 확인하세요.")
```
