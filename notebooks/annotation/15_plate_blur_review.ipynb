# 원판 블라인드 판독 · 무릎 통과에서 바 접촉까지

가까운 쪽에서 **선수가 들어 올리는 원판**을 봅니다. 흐림과 가림을 따로 기록해 나중에 어떤 조건에서 모델이 놓치는지 비교합니다. 모델 예측은 모든 판독을 마친 뒤 다음 실험에서 엽니다.

```python
from IPython.display import display
from wlcv.annotation.plate_blur import REVIEW_FILE, progress_table, review_editor, review_frames

frames = review_frames()
for video in dict.fromkeys(video for video, index in frames):
    indices = [index for name, index in frames if name == video]
    print(f"{video} · 무릎 통과 {indices[0]} → 접촉 {indices[-1]} · {len(indices)}장")
print(f"전체 {len(frames)}장 · 첫 프레임은 0번입니다.")
```

## 원본을 보고 저장하기

1. **테두리 선명도**는 가까운 원판의 바깥 테두리에서 판독합니다. 보이는 부분에서 원판이 끝나는 위치를 따라갈 수 있으면 `선명`, 바깥 경계 자체가 이동 방향으로 늘어나거나 겹쳐 끝 위치가 모호하면 `움직임으로 흐림`을 선택합니다. 가림·압축·초점 때문에 구분하기 어려우면 `판단 어려움`을 선택합니다. 내부 글자와 허브가 번져도 바깥 테두리는 선명할 수 있습니다.
2. **원판 가림**은 별도로 고릅니다. 몸이나 다른 장비가 원판 일부를 덮으면 `가림 있음`입니다. 원판이 흐리면서 동시에 가려질 수 있습니다.
3. **허브 중심**은 바벨 축이 원판을 통과하는 한 점입니다. 보이면 원본을 클릭하고, 확신이 없으면 `판독 불가`를 고릅니다. 잘못 찍으면 다시 클릭하세요. b02 34·35·36번은 이전에 예측 없이 찍은 점을 재사용할 수 있습니다. 그 좌표는 원본 위에 표시하지 않습니다.
4. **저장하고 다음**을 누르면 CSV에 저장됩니다. 프레임 목록에서 저장된 장을 다시 열어 고칠 수 있습니다. 버튼을 누르기 전에 다른 장으로 이동하면 그 장의 미저장 입력은 사라집니다. 중간에 닫아도 셀을 다시 실행하면 첫 미판독 장에서 이어집니다.

```python
display(review_editor(frames))
```

## 저장 확인

아래 셀을 다시 실행하면 선수별 진행률을 확인할 수 있습니다. `판독 불가`도 한 장을 판독한 것으로 셉니다. 허브 위치 오차를 계산할 때는 직접 찍은 좌표만 사용합니다.

```python
progress = progress_table(frames)
display(progress)
saved = int(progress["저장"].sum())
print(f"{saved}/{len(frames)}장 저장 · {REVIEW_FILE}")
if saved == len(frames):
    print("원본 판독 완료. 다음은 같은 best.pt의 640·800 추론 비교입니다.")
```
